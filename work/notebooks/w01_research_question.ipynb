## 2. The question: decision, action, cost of a wrong call

# Research Question

## Search Question

Which content pages should be reviewed first for refresh or optimization based on observable search and engagement signals?

## Unit of Analysis

One content page.

## Output

A ranked content opportunity score.

## Decision

Which pages should be reviewed first by a content reviewer?

## Action

A reviewer may:

- Refresh outdated content
- Expand thin content
- Investigate declining content
- Improve engagement
- Monitor pages showing risk signals

## Cost of a Wrong Recommendation

### False Positive

A page is recommended even though it does not require attention. This wastes reviewer time and resources.

### False Negative

A page with meaningful opportunity is missed. This results in lost opportunities to improve traffic, engagement, or search visibility.

## Why Data and ML May Help

Content performance is influenced by multiple factors such as impressions, clicks, freshness, age, engagement, and search visibility. Machine learning may improve prioritization by detecting interactions among these signals that simple rules miss.

## 3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Unique content pages:", df["content_id"].nunique())
print("Declining pages:", (df["trend_direction"]=="down").sum())
print("Declining percentage:",
      round((df["trend_direction"]=="down").mean()*100,2), "%")

print("\nAverage impressions:",
      round(df["impressions_90d"].mean(),2))

print("Average sessions:",
      round(df["sessions_90d"].mean(),2))

print("Average CTR:",
      round(df["ctr"].mean(),4))
```
